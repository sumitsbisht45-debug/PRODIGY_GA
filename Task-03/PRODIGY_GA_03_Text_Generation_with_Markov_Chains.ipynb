```python
# Task 01 - Text Generation with Markov Chains

import random
import re
from collections import defaultdict, Counter

print("Libraries imported successfully!")
```

```python
# Step 2: Create the training text dataset

text = """
Artificial intelligence is changing the way people work and learn.
Machine learning helps computers learn patterns from data.
Generative AI can create text, images, and other types of content.
Technology is developing rapidly and creating new opportunities.
Students can use programming and artificial intelligence to solve problems.
Learning new technologies requires practice, patience, and consistency.
Python is a popular programming language used in artificial intelligence.
Data is an important part of building intelligent systems.
Artificial intelligence can help people automate repetitive tasks.
The future of technology depends on creativity, innovation, and responsible development.
"""

print("Dataset loaded successfully!")
print("Number of characters:", len(text))
print("\nSample text:")
print(text[:300])
```

```python
# Step 3: Preprocess the text

# Convert text to lowercase
clean_text = text.lower()

# Remove unnecessary characters
clean_text = re.sub(r'[^a-zA-Z0-9\s]', '', clean_text)

# Split text into words
words = clean_text.split()

print("Text preprocessing completed successfully!")
print("Total number of words:", len(words))
print("\nFirst 30 words:")
print(words[:30])
```

```python
# Step 4: Build the Markov Chain model

markov_chain = defaultdict(Counter)

# Count which word comes after each word
for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]
    markov_chain[current_word][next_word] += 1

print("Markov Chain model created successfully!")
print("Number of states:", len(markov_chain))

# Display a few learned transitions
print("\nSample learned transitions:")

for word in list(markov_chain.keys())[:10]:
    print(word, "->", dict(markov_chain[word]))
```

```python
# Step 5: Create a text generation function

def generate_text(start_word, length=30):
    current_word = start_word.lower()
    generated_words = [current_word]

    for _ in range(length - 1):
        if current_word not in markov_chain:
            break

        possible_words = list(markov_chain[current_word].keys())
        frequencies = list(markov_chain[current_word].values())

        # Randomly select the next word based on learned frequencies
        current_word = random.choices(
            possible_words,
            weights=frequencies,
            k=1
        )[0]

        generated_words.append(current_word)

    return " ".join(generated_words)


print("Text generation function created successfully!")
```

```python
# Step 6: Generate text using the Markov Chain

start_word = "artificial"

generated_text = generate_text(start_word, length=30)

print("Starting word:", start_word)
print("\nGenerated Text:")
print(generated_text)
```

```python
# Step 7: Generate text using different starting words

starting_words = ["artificial", "machine", "technology", "students", "python"]

for start_word in starting_words:
    print("=" * 60)
    print("Starting Word:", start_word)
    print("Generated Text:")
    print(generate_text(start_word, length=30))
    print()
```

```python
# Step 8: Save generated text to a file

results = []

for start_word in starting_words:
    generated = generate_text(start_word, length=30)

    results.append(
        f"Starting Word: {start_word}\n"
        f"Generated Text: {generated}\n"
    )

with open("generated_text.txt", "w", encoding="utf-8") as file:
    file.write("\n" + "=" * 70 + "\n".join(results))

print("Generated text saved successfully!")
print("File name: generated_text.txt")
```

```python
# Step 9: Display the final results and project summary

print("=" * 70)
print("TASK 03 - TEXT GENERATION WITH MARKOV CHAINS")
print("=" * 70)

print("\nModel Type: Word-level Markov Chain")
print("Training Dataset: Custom text corpus")
print("Number of Starting Words Tested:", len(starting_words))
print("Maximum Generated Words per Sample: 30")

print("\nStarting Words Tested:")
for word in starting_words:
    print("-", word)

print("\nTask completed successfully!")
print("The Markov Chain generated text based on the learned word transitions.")
print("Generated results have been saved in generated_text.txt")
```
